`````python
import os
from dotenv import load_dotenv, find_dotenv
from huggingface_hub import InferenceClient

# Load environment variables
#_ = load_dotenv(find_dotenv())

# Get Hugging Face API token
API_token = os.getenv("my api")

# Create client
client = InferenceClient(
    model="meta-llama/Llama-3.1-8B-Instruct",
    token= API_token
)
`````

`````python
# helper function
def get_completion(prompt):
    response = client.chat_completion(
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0
    )

    return response.choices[0].message.content
`````

# Prompting Principles

## Principle 1: Write Clear and Specific Instructions

Provide explicit and unambiguous instructions to help the model understand exactly what you want.

## Principle 2: Give the Model Time to Think

Encourage the model to reason through complex tasks step by step before generating a final answer.

---

# Tactics

## Tactic 1: Use Delimiters to Clearly Indicate Distinct Parts of the Input

Delimiters help separate instructions, context, examples, and input data, making prompts easier for the model to interpret.

### Common Delimiters

- Triple backticks: ```` ``` ````
- Triple quotes: `"""`
- Angle brackets: `< >`
- XML tags: `<tag> </tag>`
- Colons: `:`

### Example

```text
Summarize the text delimited by triple backticks into a single sentence.

```text
Prompt engineering is the process of designing effective prompts
for large language models.
```
```

`````python
text = f"""
You should express what you want a model to do by 
providing instructions that are as clear and  
specific as you can possibly make them. 
This will guide the model towards the desired output, 
and reduce the chances of receiving irrelevant  
or incorrect responses. Don't confuse writing a  
clear prompt with writing a short prompt.  
In many cases, longer prompts provide more clarity 
and context for the model, which can lead to 
more detailed and relevant outputs.
"""
prompt = f"""
Summarize the text delimited by triple backticks  
into a single sentence.
```{text}```
"""
response = get_completion(prompt)
print(response)
`````

- Reduces ambiguity in responses

- HTML
- XML
- Markdown
- CSV
- Tables

## Tactic 3: Ask the Model to Check Whether Conditions Are Satisfied

Before performing a task, instruct the model to verify whether specific conditions or requirements are met. If the conditions are not satisfied, the model should indicate this instead of attempting to complete the task incorrectly.

### Benefits

- Reduces incorrect outputs
- Improves reliability
- Helps validate input data
- Prevents the model from making unsupported assumptions

### Example

**Prompt**

```text
You will be provided with text delimited by triple backticks.

If it contains a sequence of instructions, rewrite those instructions in the following format:

Step 1 - ...
Step 2 - ...
Step 3 - ...

If the text does not contain a sequence of instructions, simply write:

"No steps provided."

```{text}```
```

### Input with Instructions

```text
Making a cup of tea:

1. Boil water.
2. Add tea leaves.
3. Pour water into the cup.
4. Let it steep for 3 minutes.
```

### Output

```text
Step 1 - Boil water.
Step 2 - Add tea leaves.
Step 3 - Pour water into the cup.
Step 4 - Let it steep for 3 minutes.
```

### Input Without Instructions

```text
Tea is one of the most popular beverages in the world.
```

### Output

```text
No steps provided.
```

### Best Practice

Ask the model to:

1. Check whether a condition is true.
2. Explain the result if needed.
3. Proceed only when the condition is satisfied.

**Template**

```text
Check whether the following text satisfies the specified condition.

Condition:
<condition>

Text:
```<text>```

If the condition is satisfied:
<perform task>

Otherwise:
<return alternative response>
```

`````python
text_1 = f"""
Making a cup of tea is easy! First, you need to get some 
water boiling. While that's happening, 
grab a cup and put a tea bag in it. Once the water is 
hot enough, just pour it over the tea bag. 
Let it sit for a bit so the tea can steep. After a  
few minutes, take out the tea bag. If you  
like, you can add some sugar or milk to taste.  
And that's it! You've got yourself a delicious  
cup of tea to enjoy.
"""
prompt = f"""
You will be provided with text delimited by triple quotes. 
If it contains a sequence of instructions,  
re-write those instructions in the following format:

Step 1 - ...
Step 2 - …
…
Step N - …

If the text does not contain a sequence of instructions,  
then simply write \"No steps provided.\"

\"\"\"{text_1}\"\"\"
"""
response = get_completion(prompt)
print("Completion for Text 1:")
print(response)
`````

## Tactic 4: Few-Shot Prompting

Few-shot prompting involves providing the model with a few examples of the desired input-output behavior before asking it to perform the task. These examples help the model understand the pattern, format, tone, or reasoning style you expect.

### Benefits

- Improves response consistency
- Guides the model toward the desired format
- Helps with complex or specialized tasks
- Reduces ambiguity

### Example

**Prompt**

```text
Your task is to answer in a consistent style.

Child: Teach me about patience.

Grandparent: Patience is like planting a seed. You water it and care for it, and over time it grows into something beautiful.

Child: Teach me about courage.

Grandparent: Courage is like walking through a dark forest. It can feel scary, but each step forward helps you discover your strength.

Child: Teach me about perseverance.
```

### Output

```text
Grandparent: Perseverance is like climbing a mountain. The journey can be difficult and tiring, but each step brings you closer to the top and teaches you the value of determination.
```

### Another Example

**Prompt**

```text
Translate English to French.

English: Hello
French: Bonjour

English: Thank you
French: Merci

English: Good morning
French:
```

### Output

```text
Bonjour
```

### Best Practice

1. Provide 2–5 high-quality examples.
2. Ensure examples follow the exact format you want.
3. Use representative examples that cover common cases.
4. Keep examples consistent in style and structure.

### Template

```text
Input: Example 1
Output: Desired Output 1

Input: Example 2
Output: Desired Output 2

Input: Example 3
Output: Desired Output 3

Input: New Input
Output:
```

Few-shot prompting is especially useful when zero-shot prompting does not produce reliable results or when a specific response style, format, or behavior is required.

`````python
prompt = f"""
Your task is to answer in a consistent style.

<child>: Teach me about patience.

<grandparent>: The river that carves the deepest 
valley flows from a modest spring; the 
grandest symphony originates from a single note;  
the most intricate tapestry begins with a solitary thread.

<child>: Teach me about resilience.
"""
response = get_completion(prompt)
print(response)
`````

# Principle 2: Give the Model Time to Think

Encourage the model to reason through a problem step by step before generating a final answer. Breaking complex tasks into smaller steps often improves accuracy and reduces reasoning errors.

## Tactic 1: Specify the Steps Required to Complete a Task

Explicitly instruct the model to follow a sequence of steps when solving a problem. This helps the model organize its reasoning process and produce more reliable results.

### Benefits

- Improves reasoning quality
- Reduces mistakes in complex tasks
- Produces more structured outputs
- Makes responses easier to verify

### Example

**Prompt**

```text

1. Summarize the text delimited by triple backticks in one sentence.
2. Translate the summary into French.
3. List each language in the translation.
4. Output a JSON object with the following keys:
   summary
   translation
   num_languages

Text:
```Large language models are transforming how people interact with technology.```
```

### Output

```json
{
  "summary": "Large language models are transforming human interaction with technology.",
  "translation": "Les grands modèles de langage transforment l'interaction humaine avec la technologie.",
  "num_languages": 1

```

### General Template

```text
Perform the following actions:

1. <Step 1>
2. <Step 2>
3. <Step 3>
4. <Step 4>

Input:
```<text>```
```

### Best Practice

- Break complex tasks into smaller steps.
- Specify the order in which steps should be completed.
- Clearly define the desired output format.
- Use numbered instructions whenever possible.

By explicitly outlining the required steps, you give the model time to process information methodically and generate more accurate responses.

`````python
text = f"""
In a charming village, siblings Jack and Jill set out on  
a quest to fetch water from a hilltop  
well. As they climbed, singing joyfully, misfortune  
struck—Jack tripped on a stone and tumbled  
down the hill, with Jill following suit.  
Though slightly battered, the pair returned home to 
comforting embraces. Despite the mishap,  
their adventurous spirits remained undimmed, and they  
continued exploring with delight.
"""
# example 1
prompt_1 = f"""
Perform the following actions: 
1 - Summarize the following text delimited by triple 
backticks with 1 sentence.
2 - Translate the summary into French.
3 - List each name in the French summary.
4 - Output a json object that contains the following 
keys: french_summary, num_names.

Separate your answers with line breaks.

Text:
```{text}```
"""
response = get_completion(prompt_1)
print("Completion for prompt 1:")
print(response)
`````

## Tactic 2: Instruct the Model to Work Out Its Own Solution Before Rushing to a Conclusion

When evaluating reasoning or solving complex problems, ask the model to derive its own solution first rather than immediately accepting a proposed answer. This helps reduce reasoning errors and improves the quality of the final response.

### Benefits

- Improves logical reasoning
- Reduces incorrect conclusions
- Encourages step-by-step analysis
- Helps identify mistakes in provided solutions

### Example

**Prompt**

```text
Determine whether the student's solution is correct.

Problem:
I'm building a solar power installation and need help working out the financials.

- Land costs $100/square foot
- I can buy solar panels for $250/square foot
- I negotiated a contract for maintenance that will cost me a fixed $100k per year and an additional $10/square foot
- What is the total cost for the first year of operations as a function of the number of square feet?

Student's Solution:
Let x be the size of the installation in square feet.

Costs:
- Land cost: 100x
- Solar panel cost: 250x
- Maintenance cost: 100,000 + 10x

Total cost:
360x + 100,000

First, work out your own solution to the problem.
Then compare your solution to the student's solution and evaluate whether it is correct.
Do not decide whether the student is correct until you have completed your own reasoning.
```

### Model Reasoning

1. Land cost = 100x
2. Solar panel cost = 250x
3. Maintenance cost = 100,000 + 10x
4. Total cost = 100x + 250x + (100,000 + 10x)
5. Total cost = 360x + 100,000

### Output

```text
The student's solution is correct.

The total first-year cost is:

360x + 100,000

The student's calculations match the independently derived solution.
```

### General Template

```text
Problem:
<problem statement>

Proposed Solution:
<solution>

Instructions:

1. Work out your own solution.
2. Show your reasoning step by step.
3. Compare your solution with the proposed solution.
4. Determine whether the proposed solution is correct.
5. Explain any differences.
```

### Best Practice

- Ask the model to derive its own answer first.
- Separate reasoning from evaluation.
- Delay judgment until after analysis.
- Use for math, logic, coding, and fact-checking tasks.

This tactic is especially useful when verifying answers, reviewing code, evaluating solutions, or checking reasoning in RAG and Agent-based systems.

# Model Limitations: Hallucinations

A hallucination occurs when a language model generates information that sounds plausible and confident but is actually incorrect, fabricated, or unsupported by the provided information.

## What is a Hallucination?

Large Language Models (LLMs) do not inherently know whether information is true or false. They generate text based on patterns learned from training data. As a result, they may sometimes:

- Invent facts
- Create fictional product names
- Misattribute quotes
- Generate incorrect references
- Fabricate sources or details

## Example

Suppose the following prompt is provided:

```text
Tell me about the AeroGlide UltraSlim Toothbrush by Boie.
```

A model might respond:

```text
The AeroGlide UltraSlim Toothbrush is a premium toothbrush from Boie featuring ultra-soft bristles and an ergonomic design.
```

This response sounds reasonable, but it may be hallucinated.

### Reality

- Boie is a real company.
- "AeroGlide UltraSlim Toothbrush" is not a real Boie product.
- The model generated a plausible-sounding product name and description.

## Why Hallucinations Occur

LLMs are trained to predict the next most likely word, not to verify facts.

When asked about unfamiliar or nonexistent entities, models may:

1. Assume the entity exists.
2. Infer likely characteristics.
3. Generate a confident answer.

## Risks of Hallucinations

- Incorrect information
- Poor business decisions
- Misleading users
- Reduced trust in AI systems

## Mitigation Strategies

### 1. Ask the Model to Verify Information

```text
If you are uncertain whether the product exists, state that you do not know.
```

### 2. Use External Knowledge Sources

- Search engines
- Databases
- APIs
- RAG (Retrieval-Augmented Generation)

### 3. Ask for Sources

```text
Provide sources supporting your answer.
```

### 4. Ground Responses in Provided Context

```text
Answer only using the information contained in the text below.
If the answer is not contained in the text, say "I don't know."
```

## Key Takeaway

Hallucinations occur when a model generates information that appears factual but is not grounded in reality or supporting evidence. A classic example is when a model correctly recognizes a real company (Boie) but invents a nonexistent product associated with that company.
